```python
names = ["ammu","minnu","anu"]
count = 0
for name in names:
    count += name.lower().count('a')
print("name of a:",count)
```

```python
color_list_1 = ["black","brown","white","blue","green","yellow"]
color_list_2 = ["red","pink","yellow","orange"]
for color in color_list_1:
    if color not in color_list_2:
        print(color)
```

```python
d = {3: "C",2: "B",1: "A"}
print("Acsending:", dict(sorted(d.items())))
print("Desensing:", dict(sorted(d.items(),reverse = True)))
```
